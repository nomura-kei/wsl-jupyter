# Jupyter 環境

本環境は、ローカルネットワーク内での利用を想定しています。
インターネット上には公開しないでください。
公開する場合は、前段に認証機能を設けるなどの対策を実施ください。

## 環境の説明
- OS: Alpine を使用しております。
- Shell: sh(ash), bash が利用可能です。
- root 権限: パスワード無しで、sudo が利用可能です。
- wsl: wsl コマンドを利用可能です。
- その他: ツール類は、最低限必要な uv, git, curl が入っています。
  必要に応じて、インストールしてください。

```python
%%bash
# wsl.exe コマンドで、日本語文字化けが発生する場合は、 iconv で変換してください。
wsl.exe --version| iconv -f UTF-16LE -t UTF-8
```
